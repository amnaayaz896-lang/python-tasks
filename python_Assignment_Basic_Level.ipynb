```python
#Task 1: Cinema Ticket Price

user_name = str(input("Enter your name: "))
user_age = int(input("Enter your age: "))
print("Customer:", user_name)
if user_age >= 5 or user_age <= 12:
  print("Ticket price: 300")
elif user_age >= 13 or user_age <= 59:
  print("Ticket price: 700")
elif user_age >= 60 or user_age <= 100:
  print("Ticket price: 400")
else:
  print("Not Allowed")
```

```python
#Task 2 Temperature Converter

celsius_temperature = int(input("what is the celsius temperature? "))
Fahrenheit = (celsius_temperature * 9/5) + 32
print("Fahrenheit:", Fahrenheit )
if celsius_temperature < 10:
  print("Very Cold")
elif celsius_temperature <= 10 or celsius_temperature <= 19:
  print("Cold")
elif celsius_temperature <= 20 or celsius_temperature <= 29:
  print("Normal")
elif celsius_temperature >= 30:
  print("Hot")
else:
  print("Undefined")
```

```python
#Task 4 Weekly Steps Tracker

total_steps = 0

for i in range (1, 7):
  d1_st = int(input("enter your day 1 steps: "))
  d2_st = int(input("enter your day 2 steps: "))
  d3_st = int(input("enter your day 3 steps: "))
  d4_st = int(input("enter your day 4 steps: "))
  d5_st = int(input("enter your day 5 steps: "))
  d6_st = int(input("enter your day 6 steps: "))
  d7_st = int(input("enter your day 7 steps: "))
  break
total_steps = d1_st + d2_st + d3_st + d4_st + d5_st + d6_st + d7_st
avg_st = total_steps / 7 * 100
print("Total Steps:", total_steps )
print("Average Steps:", avg_st )

if avg_st >= 5000:
  print("Active user")
else:
  print("Needs more Activity")
```
