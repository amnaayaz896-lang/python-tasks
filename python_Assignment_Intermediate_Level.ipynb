## Task 10: Contact Book

### Task Description

Dictionary use karke simple contact book banayein.

### Requirements

User ko menu show karein:

```
1. Add Contact
2. Search Contact
3. Display All Contacts
4. Exit
```

Program should:

1. Contact name aur phone number store kare.
2. Name se contact search kare.
3. Duplicate contact name check kare.
4. Tamam contacts display kare.
5. Exit select hone tak program run karta rahe.

### Suggested Dictionary

```

}
```

````python
#Task 10 Contact Book

contact_info = {}

def add_contact():
   contact_number = int(input("Enter contact number: "))
   for contact in contact_info:
    if contact{"number"} == contact_info:
      print("contact number already exist")
      return

  def search_contact():
    contact_name = str(input("Enter contat name: "))
    for contact in contact_info:
      if contact{"name"} == contact_info:
        print("contact name already exist")
        return

  contact_info.append({
      "number" : contact_number
      "name" : contact_name
  })

  def display_contact():
    print("\n==== Display Contact ====")
    for contact in contact_info:
      print(f'Contact Name: {contact["name"]}')
      print(f'Contact Number: {contact["number"]}')

while True:
  print("1.Add Contact")
  print("2. Search Contact")
  print("3. Display All Contact")
  print("4. Exit")

choice = int(input("Enter your choice: "))

if choice == 1:
  add_contact()
elif choice == 2:
  search_contact()
elif choice == 3:
  display_contact()
elif choice == 4:
  print("Exit")
 break
else:
  print("Invalid Choice")

````
