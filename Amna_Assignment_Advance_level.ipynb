## Task 14: Restaurant Ordering System

### Task Description

Customer restaurant menu se multiple food items order kar sake aur program complete bill generate kare.

### Example Menu

```
menu= {"Burger":500,"Pizza":1200,"Pasta":800,"Drink":200
}
```

### Requirements

1. Restaurant menu display karein.
2. Customer se item name aur quantity input lein.
3. Check karein ke item menu mein available hai.
4. Multiple items order karne ki facility dein.
5. Har item ka subtotal calculate karein.
6. Order ko list of dictionaries mein store karein.
7. Total bill calculate karein.
8. Total bill par 13% tax calculate karein.
9. Rs. 3,000 ya above par 10% discount dein.
10. Function use karke formatted receipt display karein.


````python
#Task 14 Restaurant Ordering System

menu= {
    "Burger" : 5000,
     "Pizza" : 1200,
     "Pasta" : 800,
     "Drink" : 200
}
orders = []
print("===== Restaurant Menu =====")
for key, value in menu.items():
  print(f"{key}:{value}")

while True:
  item_name = str(input("Enter your item name: "))
  if item_name in menu:
    quantity = int(input("Enter quantity of your item: "))
    price = menu[item_name]
    sub_total = price * quantity

    orders.append({
        "item": item_name,
        "quantity": quantity,
        "sub_total": sub_total
    })
  else:
    print("This is not available")
    choice = str(input("Do you want to add more item? Yes/No: "))
    if choice == "No":
      break

  total= 0
for order in orders:
 total = total + order["sub_total"]

tax = (total/100) * 13
if total >= 3000:
 discount = (total/100) * 10

else:
  discount = 0

grand_total = total + tax - discount

print("\nRECEIPT")
def receipt():

 for order in orders:
  print(f"{order["item"]} X {order["quantity"]} = {order["sub_total"]}")

print("Total: ", total)
print("Tax: ", tax)
print("Discount: ", discount)
print("Grand Total: ", grand_total)

receipt()


````

````python
#Task 15 Clinic Appointment Manager
patients = []

def add_patient():
   patient_id = int(input("Enter your id:"))
   for patient in patients:
     if patient["id"] == patient_id:
      print("Patient already exist")
     return
   patient_name = str(input("Enter your Name: "))
   patient_age = int(input("Enter your age: "))
   problem = str(input("Enter the problem: "))
   time = int(input("Enter the time: "))
   emergency = str(input("Is this emergency patient? yes/no"))

   patients.append({
     "id": patient_id,
     "name": patient_name,
     "age ": patient_age,
     "problem": problem,
     "time": time,
     "emergency": emergency
 })

def search_patient():
   for patient in patients:
    if patient["id"] == "patient_id":
     print("Patient found")
     print(f'Patient name: {patient["name"]}')
     print(f'Patient age: {patient["age"]}')
     print(f'Patient problem: {patient["problem"]}')
     print(f'Patient time: {patient["time"]}')
     print(f'Patient emergency: {patient["emergency"]}')
   else:
    print("Patient not found")

def display_patients():
  print("\n=== Emergency Patients ===")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  print("\n=== Normal Patients ===")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f'Patient name: {patient["name"]}')
      print(f'Patient age: {patient["age"]}')
      print(f'Patient problem: {patient["problem"]}')
      print(f'Patient time: {patient["time"]}')

  def deleted_patient():
    patient_id = int(input("Enter your id: "))
    for patient in patients:
     if patient["id"] == patient_id:
      patient.remove(patient)
      print("Patient deleted")

  def count_patient():
    emergency_count = 0
    normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1

while True:
    print("1. Add Patient")
    print("2. Search Patient")
    print("3. Display Patient")
    print("4. Deleted Patient")
    print("5. Count Patient")
    print("6. Exit")


choice = int(input("Enter your choice:"))

 if choice == 1:
    add_patient()
 elif choice == 2:
    search_patient()
 elif choice == 3:
    display_patients()
 elif choice == 4:
    deleted_patient()
 elif choice == 5:
    count_patient()
 elif choice == 6:
    break
 else:
    print("Invalid choice")
````
